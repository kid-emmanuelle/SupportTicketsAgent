```python
import pandas as pd
from collections import Counter
import re
```

```python
tickets = pd.read_csv("../data/tickets_clean.csv")
```

```python
tickets
```

```python
def letter_frequency(corpus):
    counts = Counter()
    total = 0

    for text in corpus:
        text = text.lower()
        # only keep letters we care about
        letters = re.findall(r"[a-zäöüß]", text)

        for ch in letters:
            counts[ch] += 1
            total += 1

    freq = {chr(c): 0.0 for c in range(ord("a"), ord("z") + 1)}
    for ch in ["ä", "ö", "ü", "ß"]:
        freq[ch] = 0.0

    if total == 0:
        return freq

    for ch in freq:
        freq[ch] = counts.get(ch, 0) / total

    return freq
```

```python
en_tickets = list(tickets[tickets.language == "en"]["body"])
de_tickets = list(tickets[tickets.language == "de"]["body"])
```

```python
en_freq = letter_frequency(en_tickets)
de_freq = letter_frequency(de_tickets)

en_freq, de_freq
```

```python
import json

with open("../src/support_agent/freq_baseline/en_freq.json", "w") as f:
    json.dump(en_freq, f, indent=2)

with open("../src/support_agent/freq_baseline/de_freq.json", "w") as f:
    json.dump(de_freq, f, indent=2)
```
